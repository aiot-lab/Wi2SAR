## 3. Run Direction Finding (Offline Mode)

Now let's run the direction finding algorithm on the sample data using the `main` module directly.

````python
import main

# Define parameters
data_dir = './data/case5-3d'
realtime = False
calibration = True
no_plot = True  # We will plot manually in the notebook
pos_elevation = True

print("Running direction finding...")
# Run the main function and capture results
true_angles, predicted_angles = main.main(data_dir, realtime, calibration, no_plot, pos_elevation)

print(f"\nProcessed {len(true_angles)} measurements.")
````

## 4. Visualization of Results

Let's visualize the ground truth vs. predicted angles.

````python
# Extract azimuth and elevation for plotting
true_az = [loc.azimuth for loc in true_angles]
true_el = [loc.elevation for loc in true_angles]
pred_az = [loc.azimuth for loc in predicted_angles]
pred_el = [loc.elevation for loc in predicted_angles]

# Create a polar plot for Azimuth
fig = plt.figure(figsize=(12, 6))

# Subplot 1: Azimuth Comparison (Polar)
ax1 = fig.add_subplot(121, projection='polar')
ax1.set_title('Azimuth: True vs Predicted', pad=20)

# Convert to radians for polar plot
true_az_rad = np.radians(true_az)
pred_az_rad = np.radians(pred_az)

ax1.scatter(true_az_rad, np.ones_like(true_az_rad), c='green', marker='o', s=100, label='True Azimuth', alpha=0.6)
ax1.scatter(pred_az_rad, np.ones_like(pred_az_rad)*0.9, c='red', marker='x', s=100, label='Predicted Azimuth', alpha=0.8)
ax1.set_ylim(0, 1.2)
ax1.set_yticks([])  # Hide radial ticks
ax1.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))

# Subplot 2: Elevation Comparison (Scatter)
ax2 = fig.add_subplot(122)
ax2.set_title('Elevation: True vs Predicted')
ax2.scatter(range(len(true_el)), true_el, c='green', marker='o', label='True Elevation', alpha=0.6)
ax2.scatter(range(len(pred_el)), pred_el, c='red', marker='x', label='Predicted Elevation', alpha=0.8)
ax2.set_xlabel('Measurement Index')
ax2.set_ylabel('Elevation (degrees)')
ax2.set_ylim(-90, 90)
ax2.grid(True)
ax2.legend()

plt.tight_layout()
plt.show()

# Calculate and display error metrics
az_errors = np.abs(np.array(true_az) - np.array(pred_az))
az_errors = np.minimum(az_errors, 360 - az_errors)  # Handle circular wrap-around
el_errors = np.abs(np.array(true_el) - np.array(pred_el))

print(f"Mean Azimuth Error: {np.mean(az_errors):.2f}°")
print(f"Mean Elevation Error: {np.mean(el_errors):.2f}°")
````
